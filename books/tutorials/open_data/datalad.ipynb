## Purpose and Learning Objectives

**Purpose:** Using DataLad to publish and access open datasets on the Open Science Framework (OSF).

**Learning objectives:**
- Create a DataLad dataset and save files with version control
- Publish a dataset to OSF using DataLad
- Publish a human-readable dataset to OSF
- Download a DataLad dataset from OSF

## Section 1: Publish a DataLad Dataset

### Step 1: Create a DataLad dataset

```bash
datalad create my_dataset
```

### Step 2: Add and save files

Add files to the project, then save them with a commit message:

```bash
datalad save -m "added new files"
```

### Step 3: Set up OSF authentication

Generate an OSF token at Account Settings → Personal access tokens → Create token, then authenticate with DataLad:

```bash
datalad osf-credentials
```

### Step 4: Publish as a DataLad dataset on OSF

Create a sibling and push your dataset:

```bash
# Create sibling
datalad create-sibling-osf --title best-study-ever -s osf
git config --global --add datalad.extensions.load next

# Push
datalad push --to osf
```

:::note
The last steps create a DataLad dataset on OSF, which is not easily human-readable. For a human-readable version (without the ability to download as a DataLad dataset later), see the next section.
:::

## Section 2: Publish as a Human-Readable Dataset

For a human-readable version that can be accessed through the OSF web interface:

```bash
datalad create-sibling-osf --title best-study-ever-human-readable --mode exportonly -s osf-export
git-annex export HEAD --to osf-export-storage
```

## Section 3: Download a DataLad Dataset from OSF

Download a dataset from OSF (requires it to have been uploaded as a DataLad dataset):

```bash
# Clone the dataset
datalad clone osf://ehnwz
cd ehnwz

# Download the files you need
datalad get .
```

## Summary

You have successfully:
- Created a DataLad dataset with version control
- Published your dataset to OSF in both DataLad and human-readable formats
- Downloaded an existing OSF dataset using DataLad
- Set up OSF authentication for publishing
