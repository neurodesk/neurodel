## Purpose and Learning Objectives

**Purpose:** Using osfclient to publish and access open datasets on the Open Science Framework (OSF).

**Learning objectives:**
- Set up OSF authentication with a personal access token
- Upload a dataset to OSF using osfclient
- Download a dataset from OSF using osfclient

## Prerequisites

- Running Neurodesk
- OSF account
- pip/osfclient installed

## Section 1: Set Up an OSF Token

### Step 1: Generate a token

Generate an OSF token under your [user settings](https://osf.io/settings/tokens), then set it as an environment variable:

```bash
export OSF_TOKEN=YOURTOKEN
```

:::note
If you created your OSF account via ORCID, use the TOKEN as the username when running `osf init`. No password is required.
:::

## Section 2: Publish a Dataset

### Step 1: Initialize OSF project directory

```bash
cd /path/to/dataset
osf init
```

Enter your OSF credentials and project ID when prompted.

### Step 2: Upload data

Copy data into the directory, then upload:

```bash
osf upload -r ./data osfstorage/data
```

:::warning
Hidden files (e.g. `.DS_Store`) may cause issues. Delete them before uploading.
:::

## Section 3: Download a Dataset

### Step 1: Clone an OSF project

```bash
osf -p PROJECTID_HERE_eg_y5cq9 clone .
```

Replace `PROJECTID_HERE_eg_y5cq9` with the OSF project ID from the URL of the project (e.g., `https://osf.io/y5cq9/`).
