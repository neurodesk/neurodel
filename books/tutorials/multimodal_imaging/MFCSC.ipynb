## Purpose and Learning Objectives

**Purpose:** Tutorial for using MFCSC to calculate the mismatch between functional and structural brain connectomes.

**Learning objectives:**
- Download MFCSC input data from OSF
- Run the MFCSC tool from the Neurodesk application menu
- Interpret the mFCSC (mismatch functional-structural connectivity) output
- Understand how to adapt the workflow for MEG + dMRI data

## Prerequisites

- Running Neurodesk
- OSF account (for downloading data)
- Familiarity with connectome concepts

## Section 1: Tutorial for fMRI and dMRI Connectomes

### Step 1: Download input data from OSF

Download the "input" folder from the OSF repository: https://osf.io/d7j9n/files/osfstorage

### Step 2: Launch MFCSC

Launch mfcsc from either:
- Neurodesk → Diffusion Imaging → mfcsc → mfcsc 1.1
- Neurodesk → Functional Imaging → mfcsc → mfcsc 1.1

### Step 3: Run MFCSC

Run the following command (replace paths as needed):

```bash
mfcsc input/FC_SC_list.txt input/FC input/SC outputdir
```

Where:
- `input` = directory where input data was downloaded
- `outputdir` = directory for output

### Step 4: Examine outputs

After MFCSC finishes, `outputdir` content should match the "output" folder at https://osf.io/d7j9n/files/osfstorage - contains mFCSC connectomes encoding the mismatch between functional and structural connectivity.

## Section 2: Tutorial for MEG and dMRI Connectomes

### Step 1: Generate functional and structural connectomes

1. Use MEG connectivity tutorial to generate functional connectomes from MEG data using MNE
2. Use structural connectivity tutorial to generate structural connectomes from dMRI using MRtrix

### Step 2: Organize input data

1. Launch mfcsc from the application menu
2. Copy MEG connectomes into `input/MEG` and structural connectomes into `input/SC`
3. Create `input/MEG_SC_list.txt` listing the pairing between MEG and structural connectomes

### Step 3: Run MFCSC with MEG data

```bash
mfcsc input/MEG_SC_list.txt input/MEG input/SC outputdir
```

### Step 4: Interpret outputs

Output: connectomes encoding the mismatch between MEG and structural connectivity.
