## Purpose and Learning Objectives

**Purpose:** Guide to creating persistent conda environments on Neurodesk for Python and R analysis.

**Learning objectives:**
- Understand why the default conda environment is not persistent in Neurodesk
- Create a persistent Python conda environment with ipykernel
- Create a persistent R conda environment
- Use a custom conda environment as a Jupyter kernel

## Important Note: Persistent Environments

:::note
The default conda environment is **not persistent** across sessions. Any packages installed in the standard environment will disappear after restarting JupyterLab. To create persistent environments, store them in your home directory following the steps below.
:::

## Section 1: Create a Conda Environment

### Step 1: Open a terminal

Open a terminal from the JupyterLab Launcher.

### Step 2: Create a Python environment

For Python (using mamba or conda — both work):

```bash
mamba create -n myenv ipykernel
```

Or using conda:

```bash
conda create -n myenv ipykernel
```

### Step 3: Create an R environment

For R:

```bash
mamba create -n r_env r-irkernel
```

Or using conda:

```bash
conda create -n r_env r-irkernel
```

:::important
For Python environments, you must include `ipykernel` (or a Python version like `python=3.8`) — a kernel is required. If you forgot, add it with:
```bash
conda install ipykernel
```
:::

## Section 2: Check and Activate Environments

### Step 1: List environments

```bash
mamba env list
```

This shows all available conda environments.

### Step 2: Activate an environment

```bash
conda activate myenv
```

### Step 3: Install packages

Once activated, install packages with pip or conda:

```bash
pip install -r requirements.txt
```

## Section 3: Use the Environment as a Jupyter Kernel

### Step 1: Check available kernels

After creating the environment, a new Notebook option for that kernel will appear in the JupyterLab Launcher.

### Step 2: Switch kernels in an existing notebook

Switch kernels in an existing notebook from the top-right corner dropdown menu.

### Step 3: Select your custom environment

Your conda environment (e.g., "myenv" or "r_env") will appear as an available kernel option in both the Launcher and the kernel selector dropdown.
