## Purpose and Learning Objectives

**Purpose:** Guide to setting up a MATLAB license on Neurodesk and calling Neurodesk tools from within MATLAB.

**Learning objectives:**
- Configure MATLAB using an institutional network license or MathWorks account
- Launch the MATLAB GUI from the Neurodesk application menu
- Call Neurodesk neuroimaging tools from within MATLAB scripts
- Change MATLAB keyboard shortcuts to Windows defaults

## Prerequisites

- Running Neurodesk
- Valid MATLAB license (institutional network or MathWorks account)
- Familiarity with basic MATLAB usage

## Section 1a: Institutional Network License

### Step 1: Create license file

Run this command once, replacing the address and license number:

```bash
mkdir -p /home/jovyan/Downloads && echo -e "SERVER rtlicense1.university.edu D1234560F6 27007\nUSE_SERVER" > /home/jovyan/Downloads/network.lic
```

This creates a license file that connects to your institution's license server.

## Section 1b: MathWorks Account License

### Step 1: Open MATLAB

Navigate to: **Neurodesk → Programming → matlab → matlabGUI 2022a**

### Step 2: Select activation method

Select "Activate automatically using the internet" → Next

### Step 3: Enter credentials

Enter MathWorks account credentials (email + password, or university credentials if your institution provides a license).

### Step 4: Select license

Select the appropriate license → Next

### Step 5: Confirm activation

Keep default login name → Next → Confirm

Your MATLAB license is now activated.

## Section 2: Calling Neurodesk Tools from MATLAB

### Step 1: Add Neurodesk container to path

Add the Neurodesk container to MATLAB's execution path. For FSL:

```matlab
setenv("PATH", getenv("PATH") + ":/cvmfs/neurodesk.ardc.edu.au/containers/fsl_6.0.7.4_20231005");
```

### Step 2: Call Neurodesk tools

Then call FSL tools directly in MATLAB scripts (e.g., `fslmaths`).

### Step 3: Check available versions

:::note
The path version (`fsl_6.0.7.4_20231005`) should match the version of FSL you need. Check available versions with `ml avail` in a terminal.
:::

## Section 3: Change Keyboard Shortcuts

### Step 1: Open Preferences

Open Preferences from the MATLAB Home tab: **Home → Preferences**

### Step 2: Navigate to Keyboard settings

Navigate to: **Keyboard → Shortcuts**

### Step 3: Change shortcut scheme

Change "Emacs Default Set" to "Windows Default Set"

By default, MATLAB uses Emacs keyboard shortcuts on Linux. This change applies Windows shortcuts instead.
