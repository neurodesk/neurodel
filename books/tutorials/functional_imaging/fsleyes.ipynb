## Purpose

**FSLeyes** is the primary viewer for FSL data and one of the most widely used MRI visualisation tools in neuroimaging. It provides an intuitive interface for inspecting anatomical images, statistical maps, and brain masks in three orthogonal planes (axial, coronal, sagittal).

This tutorial covers the core FSLeyes workflow inside Neurodesk: opening images, navigating between slices, adding and configuring overlays, adjusting colormaps and transparency, and extracting voxel information.

:::{admonition} Learning Objectives
:class: tip
By the end of this tutorial you will be able to:
- Open FSLeyes from the Neurodesk application menu and terminal
- Load NIfTI images and navigate in orthographic view
- Add and configure overlay images (activation maps, brain masks)
- Adjust colormaps, ranges, and transparency for optimal visualisation
- Use the information panel to read voxel values and anatomical coordinates
- Save screenshots and export views
:::

## Step 1: Download Sample Data

Open **LXTerminal** from the Neurodesktop taskbar and run the following commands to get all data needed for this tutorial:

:::{tip}
To paste in Neurodesktop, use right-click with your mouse!
:::

```bash
# Load FSL (needed for the MNI template path)
ml fsl/6.0.7.8

# T1w anatomical image from OpenNeuro ds000102 (used for navigation sections)
cd ~/neurodesktop-storage/
datalad install https://github.com/OpenNeuroDatasets/ds000102.git
cd ds000102
datalad get sub-08/anat/sub-08_T1w.nii.gz

# MNI152 T1 2mm template from FSL (already installed, no download needed)
cp $FSLDIR/data/standard/MNI152_T1_2mm_brain.nii.gz ~/neurodesktop-storage/

# Z-stat map from NeuroVault (used for the statistical maps section)
# Steele et al. 2013 - Stroop task, young adults
wget -O ~/neurodesktop-storage/zstat-young.nii.gz \
  https://neurovault.org/media/images/77/zstat-young.nii.gz
```

## Step 2: Open FSLeyes with the T1w Image

Open FSLeyes with the anatomical image:

```bash
fsleyes ~/neurodesktop-storage/ds000102/sub-08/anat/sub-08_T1w.nii.gz &
```

The `&` runs FSLeyes in the background so the terminal stays usable.

Alternatively, open FSLeyes from the application menu (**Applications → Neurodesk → Structural Imaging → FSL → fsleyes GUI**) and load the file via **File → Add from file**.

## Navigating the Orthographic View

FSLeyes shows three orthogonal slices simultaneously:

- **Top left**: Sagittal view (side-to-side)
- **Top right**: Coronal view (front-to-back)
- **Bottom left**: Axial view (top-to-bottom)

A white crosshair marks the current cursor position in all three planes.

### Navigation Methods

1. **Click to set position**: Click anywhere in any panel to move the crosshair to that location
2. **Scroll wheel to change slice**: Scroll up/down in any panel to move through slices
3. **Keyboard navigation**:
   - Left/Right arrow keys: Move left/right in the sagittal view
   - Up/Down arrow keys: Move up/down in the coronal view
   - Page Up/Page Down: Move forward/backward in the axial view

Try clicking on the brain in the sagittal view and watch the crosshair update in all three planes simultaneously.

![FSLeyes orthographic view navigation](/static/tutorials/functional_imaging/fsleyes/fsleyes_ortho_navigation.png)
*FSLeyes showing the three orthogonal planes with a crosshair at a selected location.*

## The Overlay List and Display Panel

At the bottom of the FSLeyes window is the **Overlay List**, which shows all loaded images. Each image can be toggled on/off and adjusted independently.

### Opening the Display Panel

To adjust the appearance of an image:

1. Click the image name in the Overlay List to select it
2. Click **Settings → Ortho View 1 → Overlay Display Stteings** to open the display settings. A simpler option pannel is displayed on the top of the window.

The Display Panel allows you to adjust:

- **Brightness**: Shift the intensity range
- **Contrast**: Change the slope of the intensity mapping
- **Min/Max range**: Set the minimum and maximum values shown
- **Colour map**: Choose how values are mapped to colours (e.g., Greyscale, Red-Yellow, Blue-Lightblue)
- **Opacity/Transparency**: Blend this image with others underneath

### Example: Adjusting a T1w Image

For a T1-weighted anatomical image:
- Start with **Greyscale** colourmap
- Adjust the **brightness** slider until the brain looks clear and white
- Leave **opacity** at 100% (fully opaque) since this is your base image

![FSLeyes overlay list and display panel](/static/tutorials/functional_imaging/fsleyes/fsleyes_display_panel.png)
*FSLeyes showing the overlay list (bottom) and display panel (right) with brightness and contrast controls.*

## Working with Statistical Maps

One of the most common uses of FSLeyes is visualising activation maps overlaid on an anatomical image. Here we use the z-statistic map downloaded in Step 1 (Stroop task, young adults) with the MNI152 T1 template as the background.

### Open both images in FSLeyes

```bash
fsleyes ~/neurodesktop-storage/MNI152_T1_2mm_brain.nii.gz \
        ~/neurodesktop-storage/zstat-young.nii.gz &
```

Both images are in MNI152 space, so they align immediately.

### Display settings for the z-stat map

1. Click the z-stat image name in the **Overlay List** to select it
2. In the **Display Panel**, set:
   - **Min**: `2.3` (standard threshold, approximately p < 0.01 uncorrected)
   - **Max**: `5.0`
   - **Colour map**: `Red-Yellow`
   - **Opacity**: `80%`

Values below the minimum are hidden (transparent), so only significant clusters appear. The gradient from dark red to bright yellow shows increasing z-scores.

![FSLeyes with z-stat map overlaid on MNI brain](/static/tutorials/functional_imaging/fsleyes/fsleyes_zstat_overlay.png)
*FSLeyes showing the Stroop z-stat map overlaid on the MNI152 T1 2mm brain with Red-Yellow colourmap and z > 2.3 threshold.*

:::{tip}
To also show the negative contrast, load the same file a second time via **File → Add from file**, set **Blue-Lightblue** colourmap with **Min** `-5` and **Max** `-2.3`. This displays both positive and negative effects simultaneously.
:::

![FSLeyes with z-stat map overlaid on MNI brain with negative values](/static/tutorials/functional_imaging/fsleyes/fsleyes_zstat_overlay2.png)
*FSLeyes showing the Stroop z-stat map overlaid on the MNI152 T1 2mm brain with Red-Yellow colourmap and z -5 > 2.3 threshold.*

## The Information Panel

FSLeyes has an **Location Panel** that displays details about the voxel under the crosshair.

### Opening the Information Panel

Click **Settings → Ortho View 1 → Location**

### Information Displayed

The panel shows:

- **Voxel coordinates** (in voxel indices, e.g., [45, 67, 89])
- **World coordinates** (in mm, native image space)
- **MNI coordinates** (in standard MNI space, if available)
- **Current voxel value** for each loaded image (intensity, z-statistic, etc.)

This is particularly useful for:
- **Reporting peak activation coordinates** in papers (use MNI coordinates)
- **Checking anatomical locations** (what region is this voxel in?)
- **Verifying data quality** (are values in the expected range?)

### Example Use Case

You've found a region of interest in a statistical map. Click on it in FSLeyes and note:
- The **voxel value** (z-statistic, p-value, or whatever your map contains)
- The **MNI coordinates** for reporting in your paper
- The **anatomical name** using the atlas labels (see next section)

![FSLeyes information panel](/static/tutorials/functional_imaging/fsleyes/fsleyes_info_panel.png)
*FSLeyes information panel showing voxel and world coordinates, along with the value at the crosshair.*

## Colourbar and Atlas Labels

### Adding a Colourbar

To show the mapping between colours and data values, add a colourbar:

1. Click **View → Add colourbars**

A legend appears showing how values are mapped to colours. This is important for presentations and papers.

### Adding Atlas Labels

FSLeyes can display anatomical brain region names by overlaying the MNI atlas. This makes it easy to report which structures are involved.

1. Click **Settings → Ortho settings**
2. Check **Show atlas labels**
3. Move the crosshair to different brain regions

As you move the crosshair, a text label appears showing the anatomical structure under the cursor (e.g., "Left Superior Temporal Gyrus", "Right Putamen").

### Reporting Results

A typical result report includes:
- The **brain region name** (from atlas labels)
- The **MNI coordinates** (from the Information Panel)
- The **statistical value** (z-statistic, p-value, etc.)
- The **effect size** or cluster size

Example: *"A significant cluster of activation was observed in the left middle temporal gyrus (MNI: -54, -42, 6, z = 4.2)."*

## Saving Screenshots

FSLeyes allows you to save the current view as a screenshot for presentations, papers, or documentation.

### Method 1: Using the GUI

1. Arrange the display as desired (position crosshair, adjust colours, enable labels)
2. Click **File → Take screenshot**
3. Choose a location and filename
4. Click **Save**

The screenshot is saved as a PNG image with all current visualisations.

### Method 2: Using the Command Line

For batch processing or reproducibility, use the `fsleyes render` command:

```bash
ml fsl/6.0.7.8
fsleyes render \
  -of screenshot.png \
  -zc 0 0 0 \
  ~/neurodesktop-storage/ds000102/sub-08/anat/sub-08_T1w.nii.gz
```

Common options:
- `-of` : Output filename (PNG format)
- `-zc X Y Z` : Crosshair position (world coordinates)
- `-s X Y Z` : Slice indices (voxel coordinates)

This is useful when you want to create consistent screenshots across multiple subjects or generate figures programmatically.
