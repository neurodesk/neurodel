## Purpose

This tutorial guides you through accessing and visualising the 7T HCP Retinotopy Dataset using Connectome Workbench on Neurodesk. You will learn how to download preprocessed retinotopic data from the BALSA database, open Connectome Workbench, load scene files, and explore individual retinotopic maps.

:::{admonition} Learning Objectives
:class: tip

After completing this tutorial, you will be able to:

- Download and access the HCP 7T Retinotopy dataset via BALSA
- Open Connectome Workbench from the Neurodesk application menu
- Load and navigate scene files
- Visualise individual retinotopic maps and switch between functional maps

:::

## Section 1: Download Data from BALSA

The HCP 7T Retinotopy dataset is hosted on the BALSA database and includes preprocessed data from 181 participants.

### Step 1: Register for HCP Open Access

1. Visit https://www.humanconnectome.org/study/hcp-young-adult/data-use-terms
2. Review and accept the HCP Open Access Data Use Terms
3. Create an account with your email

### Step 2: Register for BALSA

1. Go to https://balsa.wustl.edu/
2. Click "Register" and create an account
3. Verify your email address

![BALSA database registration page](/static/tutorials/functional_imaging/connectomeworkbench/1_balsa.png)
*The BALSA database registration page.*

### Step 3: Download the HCP 7T Retinotopy Dataset

1. Log into BALSA
2. Navigate to https://balsa.wustl.edu/study/9Zkk
3. Click "Download" to access the study data
4. Select and download `S1200_7T_Retinotopy_9Zkk.zip`

This archive contains preprocessed data from 181 participants including retinotopic maps (polar angle, eccentricity, visual field sign), curvature maps, myelin maps, and scene files for Connectome Workbench.

![BALSA study page for HCP 7T Retinotopy dataset](/static/tutorials/functional_imaging/connectomeworkbench/2_balsa.png)
*The BALSA study page showing the HCP 7T Retinotopy dataset download.*

### Step 4: Unzip the Downloaded Archive

1. Navigate to your Downloads folder
2. Right-click on `S1200_7T_Retinotopy_9Zkk.zip` and select "Extract Here"
3. Wait for the extraction to complete (this may take several minutes)

![Unzipping the downloaded scene file archive](/static/tutorials/functional_imaging/connectomeworkbench/3_unzip.png)
*Unzipping the S1200_7T_Retinotopy scene file archive.*

## Section 2: Open Connectome Workbench and Load Scene Files

### Step 1: Launch Connectome Workbench from Neurodesk

1. Click on the **Neurodesk** icon in your application menu
2. Navigate to **Functional Imaging** → **connectomeworkbench** → **connectomeworkbench 2.1.0** (or latest version)
3. This will open a terminal window

![Connectome Workbench in the Neurodesk application menu](/static/tutorials/functional_imaging/connectomeworkbench/4_appmenu.png)
*Navigating to Connectome Workbench in the Neurodesk application menu.*

### Step 2: Start wb_view

1. In the terminal that opened, type:
   ```bash
   wb_view
   ```
2. Press Enter to launch the Connectome Workbench viewer

![Terminal with wb_view command](/static/tutorials/functional_imaging/connectomeworkbench/5_wbview.png)
*Launching wb_view from the container terminal.*

### Step 3: Open the Scene File

1. In the Connectome Workbench window, click the **"Open Other"** button
2. A file browser dialog will appear

![Connectome Workbench Open Other dialog](/static/tutorials/functional_imaging/connectomeworkbench/6_openother.png)
*Clicking "Open Other" in Connectome Workbench.*

### Step 4: Navigate to Your Extracted Data

1. Use the file browser to navigate to the location where you extracted `S1200_7T_Retinotopy_9Zkk.zip`
2. Look for the folder containing the scene files

![File browser showing scene files](/static/tutorials/functional_imaging/connectomeworkbench/7_scenefile.png)
*Browsing to the scene file location.*

![Path navigation to downloaded data](/static/tutorials/functional_imaging/connectomeworkbench/8_path.png)
*Navigating to the path where the data was downloaded.*

### Step 5: Select and Load the Scene File

1. Select the first scene file (typically named with `Scene_` prefix)
2. Click "Open" to load it

![Selecting the scene file](/static/tutorials/functional_imaging/connectomeworkbench/9_file.png)
*Selecting the scene file to open.*

### Step 6: View the Default Scene

1. A **Scenes** window will appear on the right side
2. Click on the first option in the list to load the average retinotopic maps

![Scenes window with first option selected](/static/tutorials/functional_imaging/connectomeworkbench/10_scene.png)
*The Scenes window - select the first option to load the average maps.*

3. The main viewer will now display the retinotopic maps on the brain surface

![Loaded retinotopic maps in Connectome Workbench](/static/tutorials/functional_imaging/connectomeworkbench/11_loadedscene.png)
*The default average retinotopic maps loaded in Connectome Workbench.*

## Section 3: View Individual Participant Maps

Now that you have the average maps loaded, you can explore individual participant retinotopic maps.

### Step 1: Switch to Individual Data

1. At the **bottom** of the main viewer window, find the **Overlay Toolbox** panel and identify the top row that has its **On** checkbox ticked
2. In that row, click the **File** dropdown (the first dropdown, showing the currently loaded data file name)

![Selecting individual data from the dropdown](/static/tutorials/functional_imaging/connectomeworkbench/12_fileselection.png)
*The File dropdown in the Overlay Toolbox shows the currently loaded file. Click it to select a different data file.*

### Step 2: Select the Polar Angle File

1. From the dropdown menu, select `S1200_7T_Retinotopy181.All.Fit1_PolarAngle_MSMALL.32k_fs_LR.dscalar.nii`
2. This file contains the polar angle maps for all 181 participants

![Polar angle map for all participants](/static/tutorials/functional_imaging/connectomeworkbench/13_polarAngleALL.png)
*The polar angle map file selected for individual viewing.*

### Step 3: Browse Individual Maps by Index

1. Stay in the **same row** of the Overlay Toolbox and look immediately to the right of the File dropdown
2. You will find the **Map** selector — a dropdown with a small index spinner (▲▼) next to it. After switching to the `*.All.*` file in Step 2, this selector now contains 181 entries (one per participant) instead of just 1
3. Use the index spinner, or open the Map dropdown, to pick a participant. For example, set the index to "1" to view the first participant's polar angle map

![First individual polar angle map (index 1)](/static/tutorials/functional_imaging/connectomeworkbench/14_index1.png)
*The first individual polar angle map.*

### Step 4: Switch Between Participants

1. Change the index to different numbers to view different participants
2. For example, try index "5" to see the fifth participant's map

![Fifth individual polar angle map (index 5)](/static/tutorials/functional_imaging/connectomeworkbench/14_index5.png)
*Switching to the fifth individual polar angle map.*

### Step 5: Explore Different Functional Maps

1. At the **top** of the viewer (above the brain surface, not in the Overlay Toolbox), you will see multiple tabs for different functional maps
2. These typically include:
   - Polar Angle (visual field position)
   - Eccentricity (distance from the fovea)
   - Visual Field Sign (V1/V2/V3 identification)
   - Myelin (cortical myelination)

![Navigating between functional map tabs](/static/tutorials/functional_imaging/connectomeworkbench/15_tabs.png)
*Navigating between different functional map tabs (polar angle, eccentricity, etc.).*

3. Click on different tabs to switch between map types
4. Continue adjusting the participant index to explore individual differences in retinotopic organisation

:::{note}
Each tab keeps its **own** Overlay Toolbox state — switching tabs does not propagate the File/Map selection. If you move to a new tab and it is still showing the average, repeat the steps above on that tab to view individual participant data.
:::

## Summary

In this tutorial, you have:

- Registered for HCP Open Access and BALSA accounts
- Downloaded the HCP 7T Retinotopy dataset from the BALSA database
- Extracted the scene files to your local filesystem
- Opened Connectome Workbench from the Neurodesk application menu
- Loaded scene files and visualised average retinotopic maps
- Explored individual participant retinotopic maps using the data dropdown and index selector
- Navigated between different functional map types (polar angle, eccentricity, etc.)

You now have the skills to visualise high-quality retinotopic data from the HCP 7T dataset and understand individual variation in visual cortex organisation.
