## Purpose

Example workflow for the PhysIO Toolbox — model-based physiological noise correction of fMRI data using peripheral measures of respiration and cardiac pulsation.

:::{admonition} Statement of Need
:class: tip

Physiological noise (cardiac and respiratory fluctuations) can significantly confound fMRI analyses. PhysIO implements established noise models (RETROICOR, HRV, RVT) and integrates with SPM's Batch Editor for flexible, automated noise correction without a MATLAB license.

:::

:::{admonition} Learning Objectives
:class: tip

After completing this tutorial, you will be able to:

- Open the PhysIO GUI from the Neurodesk application menu
- Load example physiological data and run a basic PhysIO analysis
- Interpret the nuisance regressor output

:::

## Section 1: Download Example Data

Example data is already included in the Neurodesk container at `/opt/spm12`.

If not available, download from: https://www.tapas.tnu-zurich.com/examples_v5.0.0.zip

**Note on Windows**: `C:\neurodesktop-storage` is linked to `/neurodesktop-storage/` inside Neurodesk

## Section 2: Open the PhysIO GUI

1. Navigate to: **Neurodesk → Functional Imaging → physio → physioGUI r7771**
2. SPM will open automatically (this may take a moment)

![Opening the PhysIO GUI from the Neurodesk application menu](/static/tutorials/functional_imaging/physio/physio_screenshot1.jpg)
*Opening the PhysIO GUI in Neurodesk. SPM will load automatically.*

## Section 3: Run the PhysIO Example

**Step 1**: Select 'fMRI' from the modality selection screen.

**Step 2**: Press the "Batch Editor" button.

![SPM Batch Editor with PhysIO loaded](/static/tutorials/functional_imaging/physio/physio_screenshot2.jpg)
*The SPM Batch Editor with PhysIO. To create a new batch from scratch, use SPM → Tools → TAPAS PhysIO Toolbox from the top menu.*

**Step 3**: In the Batch Editor, change the working directory to an example folder using `CD` from the `Utils..` dropdown. For example: `/opt/spm12/examples/Philips/ECG3T/`

**Step 4**: Load an existing batch:
- File → Load Batch
- Select the MATLAB batch file, e.g., `philips_ecg3t_spm_job.m`

**Step 5**: Press the green "Play" button to run the analysis.

:::{note}
The example data includes physiological recordings for several scanner types (Philips, Siemens, GE). Navigate to any example folder and load its corresponding `*_spm_job.m` file.
:::

![Output nuisance regressor matrix from PhysIO](/static/tutorials/functional_imaging/physio/physio_screenshot3.jpg)
*The output nuisance regressor design matrix. If this figure does not appear, check the terminal for file path errors.*

## Summary

You have successfully:

- Opened PhysIO from the Neurodesk application menu
- Loaded an example physiological dataset
- Ran the RETROICOR noise model to generate nuisance regressors

These regressors can now be incorporated into your GLM analysis to account for physiological sources of noise in fMRI data.
