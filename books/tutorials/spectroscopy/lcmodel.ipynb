## Purpose

Analysing MR spectroscopy data using LCModel within Neurodesk.

## Learning Objectives

- Open LCModel from the Neurodesk application menu
- Set up LCModel using the setup script
- Load example Varian FID spectroscopy data
- Select an appropriate basis set and run LCModel fitting
- Inspect and interpret the fitted spectra results

## Citations

- **LCModel** - Provencher, S.W. (1993). Estimation of metabolite concentrations from localised in vivo proton NMR spectra. *Magnetic Resonance in Medicine*, 30(6), 672–679. https://doi.org/10.1002/mrm.1910300604

### Educational Resources

- [LCModel user manual](http://s-provencher.com/lcmodel.shtml)
- [Example dataset on Zenodo](https://zenodo.org/record/3904443/)

## Prerequisites

- Running Neurodesk
- No additional installation required - LCModel is pre-installed in Neurodesk

## Section 1: Open LCModel

Navigate to: **Applications → Spectroscopy → lcmodel → lcmodel 6.3**

## Section 2: Set Up and Launch the GUI

Open a terminal and run the setup script:

````python
setup_lcmodel.sh
lcmgui
````

## Section 3: Load Example Data

Example data is packed into the container (rat hippocampus spectra from [Zenodo](https://zenodo.org/record/3904443/)).

### Step 1: Select Data Format

Click **Varian** (data is in Varian FID format).

![LCModel GUI with Varian format selected](/static/tutorials/spectroscopy/lcmodel/01_varian_select.png)

*Selecting Varian as the input format in the LCModel GUI.*

### Step 2: Select the FID Data

Navigate to and select the FID data:

```
/opt/datasets/Spectra_hippocampus(rat)_TE02/s_20131015_03_BDL106_scan0/isise_01.fid
```

![Selecting the FID data file in LCModel](/static/tutorials/spectroscopy/lcmodel/02_fid_select.png)

*Navigating to and selecting the Varian FID file.*

### Step 3: Select the Basis Set

Click **Change BASIS** and select the basis set in:

```
/opt/datasets/Spectra_hippocampus(rat)_TE02/Control_files_Basis_set
```

![Selecting the basis set for LCModel fitting](/static/tutorials/spectroscopy/lcmodel/03_basis_set.png)

*Selecting the appropriate basis set for the spectral fitting.*

### Step 4: Run LCModel

Click **Run LCModel**.

![LCModel ready to run with data and basis set loaded](/static/tutorials/spectroscopy/lcmodel/04_run_lcmodel.png)

*LCModel loaded with data and basis set, ready to run.*

### Step 5: Confirm the Run

Confirm the LCModel analysis run.

![LCModel run confirmation dialog](/static/tutorials/spectroscopy/lcmodel/05_confirm.png)

*Confirming the LCModel analysis run.*

## Section 4: Inspect Results

Wait a couple of minutes for the fitting to complete. Close the results window to page through individual spectra.

![LCModel fitted spectra results](/static/tutorials/spectroscopy/lcmodel/06_results.png)

*The fitted spectra - the red line shows the LCModel fit to the experimental data.*

Results are also saved to `~/.lcmodel/saved/`

## Summary

Opened LCModel from the Neurodesk application menu, loaded example rat hippocampus spectroscopy data, selected an appropriate basis set, ran the LCModel fitting, and inspected the fitted spectra.
