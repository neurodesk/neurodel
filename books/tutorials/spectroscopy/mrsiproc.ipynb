## Section 2: Running MRSI Reconstruction (Part 1)

Open VSCode and create a new folder under `~/neurodesktop-storage/`:

![Opening a new folder in VSCode](/static/tutorials/spectroscopy/mrsiproc/open_folder.png)

*Creating a new working folder in VSCode within neurodesktop-storage.*

Load mrsiproc:

```python
ml mrsiproc/0.1.0
```

Create a bash script `run_mrsi_part1.sh` with your data paths. Here is a template:

```python
#!/bin/bash

ml mrsiproc/0.1.0

Part1_ProcessMRSI.sh \
-c ~/neurodesktop-storage/data/meas_test.dat \
-t ~/neurodesktop-storage/data/T1/ \
-a ~/neurodesktop-storage/data/INV1/ \
-b fid_1.300000ms.basis \
-o ~/neurodesktop-storage/mrsi_proc/ProcessResults/test_files \
-j LCModel_Control_Template.m \
-m "dreid" \
-L "L2,0.2"
```

**Key arguments:**

- `-c` — CSI data file (DAT, DICOM, or .mat)
- `-t` — T1 anatomical image (DICOM folder)
- `-a` — T1 AntiNoise image (DICOM folder)
- `-b` — Basis set file (.BASIS)
- `-o` — Output directory
- `-j` — LCModel control file template
- `-m` — Masking method (`"bet"`, `"thresh"`, `"voi"`, or `"dreid"` for 3D)

> **Note** — The `fid_1.300000ms.basis` and `LCModel_Control_Template.m` files are included with mrsiproc. To change the number of LCModel threads, edit the `LCModel_Control_Template.m` file (default: 8 threads).

Run the script:

```python
bash run_mrsi_part1.sh
```

![mrsiproc Part 1 running in terminal](/static/tutorials/spectroscopy/mrsiproc/run_mrsi.png)

*mrsiproc Part 1 running — this step can take several hours.*

**Warning** — Part 1 reconstruction and LCModel fitting can take **several hours** depending on data size and hardware. Make sure not to close the terminal while it is running.

## Section 3: Evaluating MRSI Data (Part 2)

Create a second script `run_mrsi_part2.sh`:

```python
#!/bin/bash

ml mrsiproc/0.1.0

Part2_EvaluateMRSI.sh \
-o ~/neurodesktop-storage/mrsi_proc/ProcessResults/test_files \
-S "Glu,Gln,Raw-abs-csi" \
-N Nifti \
-R \
-b
```

**Key arguments:**

- `-o` — Output directory from Part 1
- `-S` — Spectral maps to create (metabolite names)
- `-N Nifti` — Output format (Nifti or Both for Nifti + MINC)
- `-R` — Create ratio maps
- `-b` — Perform tissue segmentation

Run:

```python
bash run_mrsi_part2.sh
```
