## Purpose

**Quantitative susceptibility mapping (QSM)** is an MRI post-processing technique that estimates the underlying tissue magnetic susceptibility from gradient-echo phase data. QSM provides quantitative contrast sensitive to iron, calcium, and myelin - making it valuable for studying neurodegenerative diseases, brain development, and brain function.

This tutorial points you to the comprehensive **QSMxT** example notebook, which walks through the full QSM pipeline in Neurodesk.

:::{admonition} Learning Objectives
:class: tip
By the end of this tutorial you will be able to:
- Understand the basic principles and clinical relevance of QSM
- Run the QSMxT pipeline on gradient-echo MRI data
- Interpret quantitative susceptibility maps
:::

## How QSM fits into the phase processing pipeline

QSM builds on the techniques covered in the other phase processing tutorials:

Phase Unwrapping
: Recovers the continuous phase signal from wrapped $[-\pi, \pi]$ data (see [Unwrapping tutorial](unwrapping.ipynb))

Background Field Removal
: Separates the local tissue susceptibility field from large-scale background fields

Dipole Inversion
: Solves the inverse problem to convert the local field map into a quantitative susceptibility map

- [Phase Unwrapping tutorial](unwrapping.ipynb) - ROMEO-based unwrapping

:::
