## Purpose

**FreeSurfer** is a widely used software suite for processing and analysing structural MRI data of the brain. Its flagship pipeline, `recon-all`, performs cortical reconstruction and volumetric segmentation — including skull stripping, white matter segmentation, cortical surface extraction, parcellation, and thickness estimation.

This tutorial demonstrates how to run FreeSurfer interactively via the **Neurodesktop application menu**.

:::{admonition} Learning Objectives
:class: tip
By the end of this tutorial you will be able to:
- Set up FreeSurfer with the required license in Neurodesktop
- Run the full `recon-all` cortical reconstruction pipeline
- Understand the key outputs produced by FreeSurfer
:::

## Run FreeSurfer

```{figure} /static/tutorials/structural_imaging/freesurfer/open-freesurfer.png
:label: fig-open-freesurfer
:class: hidden

Opening FreeSurfer from the Neurodesk application menu.
```

Open [FreeSurfer from the Neurodesk application menu](#fig-open-freesurfer) (Image Segmentation → FreeSurfer).

In the FreeSurfer container terminal, set up the environment:

```bash
mkdir -p ~/neurodesktop-storage/freesurfer-output
export SUBJECTS_DIR=~/neurodesktop-storage/freesurfer-output
```

Now run the full cortical reconstruction pipeline:

```bash
recon-all -subject test-subject \
    -i ~/neurodesktop-storage/sub-01_ses-01_7T_T1w_defaced.nii.gz \
    -all
```

:::{note}
`recon-all -all` runs the complete FreeSurfer pipeline. This can take **several hours** depending on your system. You can monitor progress in the terminal output.
:::

![FreeSurfer recon-all running](/static/tutorials/structural_imaging/freesurfer/terminal_reconall.png)
*The FreeSurfer container terminal running `recon-all`.*


### Using FreeSurfer 8.0.0 or later

If you are using FreeSurfer version 8.0.0 or newer, you need to enable the deep learning modules before running `recon-all`:

```bash
export FS_ALLOW_DEEP=1
```

If running on a GPU, also ensure the following is set:

```bash
export neurodesk_singularity_opts='--nv'
```


## Inspect the results

Once `recon-all` completes, the outputs will be in `~/neurodesktop-storage/freesurfer-output/test-subject/`. Verify:

```bash
ls ~/neurodesktop-storage/freesurfer-output/test-subject/
```

Key output directories include:

`label/`
: Parcellations (Desikan-Killiany, Destrieux)

`mri/`
: Processed volumes (brain mask, segmentations, bias-corrected T1)

`stats/`
: Volumetric and surface statistics

`surf/`
: Cortical surfaces (pial, white, inflated)

You can use **FreeView** (included with FreeSurfer) to visualise the results:

```bash
freeview -v ~/neurodesktop-storage/freesurfer-output/test-subject/mri/T1.mgz \
    -f ~/neurodesktop-storage/freesurfer-output/test-subject/surf/lh.pial:edgecolor=red \
    ~/neurodesktop-storage/freesurfer-output/test-subject/surf/lh.white:edgecolor=blue
```
