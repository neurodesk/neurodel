## 1. Set Up OSF Authentication

Generate a personal access token at [osf.io/settings/tokens](https://osf.io/settings/tokens) and set it as an environment variable. This avoids storing credentials in your notebook:

```bash
export OSF_TOKEN=your_token_here
export OSF_USERNAME=your_email@example.com
```

Alternatively, set credentials in a `.osfcli.config` file in your home directory.

````python
%%bash
# Check if credentials are configured
# (Replace with your own credentials before running)
osf -p y5cq9 ls 2>/dev/null | head -10 || echo "Configure OSF credentials first - see osf.io/settings/tokens"
````

## 2. Explore an OSF Project

OSF projects have short IDs in their URL (e.g., osf.io/y5cq9 → ID is y5cq9). List the files in a public project:

````python
%%bash
# List files in a public OSF project (ru43c is the SWI demo dataset)
osf -p ru43c ls
````

## 3. Download Files from OSF

Download specific files or clone the entire project:

````python
%%bash
mkdir -p ~/neurodesktop-storage/osf_download/

# Download a specific file
osf -p ru43c fetch osfstorage/01_bids.zip \
    ~/neurodesktop-storage/osf_download/01_bids.zip

ls -lh ~/neurodesktop-storage/osf_download/
````

````python
%%bash
# Or clone the entire project (downloads all files)
osf -p ru43c clone ~/neurodesktop-storage/osf_clone/
ls ~/neurodesktop-storage/osf_clone/
````

## 4. Upload Files to Your OSF Project

To upload, you need to be authenticated (OSF_TOKEN set) and own or have write access to the project. Replace `YOURPROJECTID` with your project ID:

````python
%%bash
# Create a sample file to upload
mkdir -p ~/neurodesktop-storage/osf_upload/
echo "Sample data for upload" > ~/neurodesktop-storage/osf_upload/sample.txt
ls ~/neurodesktop-storage/osf_upload/
````

````python
%%bash
# Upload a file (requires authentication - replace YOURPROJECTID)
# osf -p YOURPROJECTID upload ~/neurodesktop-storage/osf_upload/sample.txt osfstorage/sample.txt
echo "Upload command (requires your project ID and OSF token):"
echo "  osf -p YOURPROJECTID upload <local_file> osfstorage/<remote_path>"
echo ""
echo "Verify upload:"
echo "  osf -p YOURPROJECTID ls"
````

## 5. Publish a Dataset Folder

Upload an entire folder to OSF:

````python
%%bash
# Upload a folder (requires authentication)
# osf -p YOURPROJECTID upload -r ~/neurodesktop-storage/my_dataset/ osfstorage/dataset/
echo "Folder upload command:"
echo "  osf -p YOURPROJECTID upload -r <local_folder>/ osfstorage/<remote_folder>/"
````

## 6. Practical Example: Download the SWI Demo Dataset

This is the same dataset used in the SWI and ROMEO phase tutorials:

````python
%%bash
mkdir -p ~/neurodesktop-storage/swi-demo/

# Download from OSF
osf -p ru43c fetch -f osfstorage/01_bids.zip \
    ~/neurodesktop-storage/swi-demo/01_bids.zip

ls -lh ~/neurodesktop-storage/swi-demo/
````

````python
%%bash
# Extract the data
unzip -o ~/neurodesktop-storage/swi-demo/01_bids.zip \
    -d ~/neurodesktop-storage/swi-demo/

ls ~/neurodesktop-storage/swi-demo/
````
