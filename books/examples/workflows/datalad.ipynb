## 1. Create a DataLad Dataset

DataLad manages datasets with full version control. Start by creating a new dataset in your storage folder:

````python
%%bash
cd ~/neurodesktop-storage/
datalad create my_datalad_demo
cd my_datalad_demo
ls
````

## 2. Add and Save Files

Create some example files and save them with DataLad:

````python
%%bash
cd ~/neurodesktop-storage/my_datalad_demo
echo "Subject,Age,Group" > participants.tsv
echo "sub-01,25,control" >> participants.tsv
echo "sub-02,28,patient" >> participants.tsv
datalad save -m "Add participant metadata"
git log --oneline -5
````

## 3. Check Dataset Status

View what has changed since the last save:

````python
%%bash
cd ~/neurodesktop-storage/my_datalad_demo
echo "Modified line" >> participants.tsv
datalad status
````

````python
%%bash
cd ~/neurodesktop-storage/my_datalad_demo
datalad save -m "Update participants"
datalad log --oneline
````

## 4. Publish to OSF

To publish to OSF, first generate a personal access token at https://osf.io/settings/tokens and set it as an environment variable:

```bash
export OSF_TOKEN=your_token_here
```

Then create a sibling and push:

````python
%%bash
cd ~/neurodesktop-storage/my_datalad_demo
# Replace with your OSF token
# export OSF_TOKEN=your_token_here
# datalad create-sibling-osf --title "my-demo-dataset" -s osf
# datalad push --to osf
echo "OSF publishing requires a personal access token - see https://osf.io/settings/tokens"
````

## 5. Download a Dataset from OSF

Clone a DataLad dataset that has been published to OSF. The OSF project ID is in the URL (e.g., osf.io/ehnwz → ID is ehnwz):

````python
%%bash
cd ~/neurodesktop-storage/
datalad clone osf://ehnwz osf_demo_download
cd osf_demo_download
ls
````

````python
%%bash
cd ~/neurodesktop-storage/osf_demo_download
datalad get .
ls -lh
````

## 6. Supported Repositories

DataLad works with many open data repositories:

| Repository | Install URL format |
|---|---|
| OpenNeuro | `https://github.com/OpenNeuroDatasets/<id>.git` |
| OSF | `osf://<project_id>` |
| DANDI | `https://github.com/dandisets/<id>.git` |
| GitHub | `https://github.com/<user>/<repo>.git` |
