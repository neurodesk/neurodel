## Why Persistent Environments?

:::important
By default, conda environments created inside Neurodesk are **not persistent** - they are lost when the container restarts. To persist environments, they must be stored in `~/neurodesktop-storage/`.
:::

## 1. Set Up Conda for Persistent Storage

Configure conda to store environments in the persistent storage directory.

```python
%%bash
# Check if conda/mamba is available
which conda && conda --version
which mamba && mamba --version || echo "mamba not found, using conda"
```

```python
%%bash
# Create the directory for persistent environments
mkdir -p ~/neurodesktop-storage/conda-envs/

# Configure conda to look for environments here
conda config --add envs_dirs ~/neurodesktop-storage/conda-envs/

# Verify configuration
conda config --show envs_dirs
```

## 2. Create a Python Environment

Create a Python 3.10 environment with common neuroimaging packages.

```python
%%bash
# Create a Python 3.10 environment with common neuroimaging packages
mamba create -y \
    -p ~/neurodesktop-storage/conda-envs/neuro-py310 \
    python=3.10 \
    nibabel \
    nilearn \
    matplotlib \
    pandas \
    numpy \
    jupyterlab \
    ipykernel
echo "Python environment created successfully"
```

## 3. Register as a Jupyter Kernel

To use this environment in Jupyter notebooks, register it as a kernel.

```python
%%bash
source activate ~/neurodesktop-storage/conda-envs/neuro-py310 || \
    conda activate ~/neurodesktop-storage/conda-envs/neuro-py310

python -m ipykernel install \
    --user \
    --name neuro-py310 \
    --display-name "Python 3.10 (Neuroimaging)"
echo "Kernel registered"
```

## 4. Create an R Environment

Create an R 4.3 environment with common neuroimaging R packages.

```python
%%bash
mamba create -y \
    -p ~/neurodesktop-storage/conda-envs/r-neuro \
    r-base=4.3 \
    r-ggplot2 \
    r-tidyverse \
    r-irkernel \
    r-rnifti
echo "R environment created"
```

```python
%%bash
# Register the R kernel for Jupyter
conda run -p ~/neurodesktop-storage/conda-envs/r-neuro \
    Rscript -e "IRkernel::installspec(name='r-neuro', displayname='R 4.3 (Neuroimaging)')"
echo "R kernel registered"
```

## 5. List and Manage Environments

View all available conda environments and learn how to manage them.

```python
%%bash
# List all environments (including persistent ones)
conda env list

# To activate an environment in a new terminal:
echo ""
echo "To activate in a terminal:"
echo "  conda activate ~/neurodesktop-storage/conda-envs/neuro-py310"
echo ""
echo "To remove an environment:"
echo "  conda env remove -p ~/neurodesktop-storage/conda-envs/neuro-py310"
```

## 6. Verify the Environment Works

Test that the Python environment has the expected packages installed.

```python
%%bash
conda run -p ~/neurodesktop-storage/conda-envs/neuro-py310 \
    python -c "import nibabel, nilearn, numpy; print('nibabel:', nibabel.__version__); print('nilearn:', nilearn.__version__); print('numpy:', numpy.__version__)"
```

## 7. Make Your Environment Reproducible

Export your environment specification so others can recreate it exactly.

```python
%%bash
conda env export -p ~/neurodesktop-storage/conda-envs/neuro-py310 \
    > ~/neurodesktop-storage/conda-envs/neuro-py310-environment.yml

echo "Environment exported to neuro-py310-environment.yml"
head -20 ~/neurodesktop-storage/conda-envs/neuro-py310-environment.yml
```
